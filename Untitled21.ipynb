```python
import numpy as np

arr = np.arange(1, 21)

print("Array:", arr)
print("Number of dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Datatype:", arr.dtype)
```

```python
import numpy as np

np.random.seed(42)

arr = np.random.randint(0, 51, size=(4, 5))

print("Array:")
print(arr)
print("Item size:", arr.itemsize, "bytes")
print("N bytes:", arr.nbytes, "bytes")
```

```python
import numpy as np

arr = np.arange(1, 13).reshape(3, 4)

print("Original array:")
print(arr)
print("\na. First two rows and first two columns:")
print(arr[:2, :2])
```

```python
print("\nb. Every alternate column:")
print(arr[:, ::2])
```

```python
print("\nc. Columns reversed:")
print(arr[:, ::-1])
```

```python

import numpy as np
arr = np.arange(1, 16)
parts = np.split(arr, 3)
print("Original array:", arr)
print("Part 1:", parts[0])
print("Part 2:", parts[1])
print("Part 3:", parts[2])
```
